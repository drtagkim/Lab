# PostgreSQL에 기존 테이블이 있을 때 변환하기

예를 들어 product라고 테이블이 있다고 하자. 다음 절차에 따라 SQL을 생성하여 실행한다.

## 파티셔닝 테이블 정의

```sql
CREATE TABLE product_partitioned (
    storeId float8,
    productID float8,
    productName text,
    category text,
    regDate timestamptz,
    priceDiscount float8,
    priceOriginal float8,
    weekId text
) PARTITION BY LIST (weekId);
```

파티셔닝 기준을 다음과 같이 정하고 각각의 파티션 테이블을 원하는 만큼 만들어준다.

```sql
CREATE TABLE product_2024w01 PARTITION OF product_partitioned
    FOR VALUES IN ('2024-W01');

CREATE TABLE product_2024w02 PARTITION OF product_partitioned
    FOR VALUES IN ('2024-W02');
...
```
