# AI 추론
## AI 추론이란?
- Training: AI 모델이 데이터에서 패턴과 규칙을 학습하는 과정. 대량의 입력과 정답(label)을 이용해 가중치(weight)를 조정. 예: GPT가 인터넷 텍스트, 책, 코드 등으로 수십억 번 학습
- Inference(추론): 학습이 끝난 모델이 새로운 입력에 대해 답을 생성하는 과정. 모델의 가중치는 고정되어 있고, 입력된 데이터를 기반으로 계산하여 결과를 예측. 시험장에서 문제를 풀어보는 것과 같은 느낌.
## LLM의 추론 과정 개념
1. Initial Input
  - 사용자가 입력
  - 예) 오늘 서울 동대문구의 날씨는?
1. Tokenization
  - 문장을 모델이 이해할 수 있도록 숫자 토큰으로 변환
  - "서울 동대문구" → [1032, 17], "날씨" → [2098] 처럼 단어·어절·글자 단위로 분할
1. Context Understanding
  - Transformer 모델로 토큰 간의 관계를 분석
  - 어떤 단어가 앞뒤 문맥에서 어떻게 연결되는지 계산(Attention)
1. Text Generation
  - 한 토큰씩 확률적으로 예측하고 연결하여 문장을 완성
  - Temperature(창의성) 등의 파라미터로 출력 스타일을 미세 조정
## OpenAI API 기반의 추론
>GPT는 Prompt-Response(PR) 구조를 바탕으로 LLM 추론을 수행함
- Prompt: 모델에게 주는 입력 지시문
- Response: 모델이 생성한 출력 결과

## Zero-shot / Few-shot 프롬프트 비교

- Zero-shot 프롬프트: 예시를 하나도 주지 않고 문제만 제시, 간단하고 빠름. 원하는 스타일이나 결과를 얻기는 어려울 수도 있음
- Few-shot 프롬프트: 몇 개의 예시를 제공하고 GPT가 패턴을 이해하게 함. 형식과 톤을 안정적으로 가져갈 수 있지만 프롬프트가 길어짐. 프롬프트 캐싱을 통해 비용을 아끼려는 적극적 노력이 필요할 수 있음.
